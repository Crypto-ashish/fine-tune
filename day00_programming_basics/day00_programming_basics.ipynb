## Lesson 1 — What a program is, and print()

A program is a list of instructions the computer follows from top to bottom, one line at a time. `print()` is the simplest instruction: it shows something on the screen. Anything after `#` is a comment: a note for humans that the computer ignores.

```python
print("Hello, Ashish!")
print("I am learning to fine-tune AI models.")
# This line is a comment. Python skips it.
print(2 + 3)
```

**What happened:** `print(...)` shows whatever is inside the brackets. Text must be inside quotes `"..."`; numbers and sums need no quotes, so `print(2 + 3)` shows the answer, `5`. The comment line produced no output.

🔗 **Where this appears in Day 1:** Day 1 uses `print(...)` to show the device, the parameter count and every answer. Lines starting with `#` are the explanations written inside the code.

## Lesson 2 — Variables: giving a value a name

A **variable** is a labelled box that stores a value. You create one with `=`, which means “store the value on the right in the name on the left”. Later you use the name instead of retyping the value.

```python
bakery = "Ovenroma"
cookies_per_tray = 24
trays = 5

print(bakery)
print(cookies_per_tray * trays)
```

**What happened:** Three boxes are created: `bakery`, `cookies_per_tray` and `trays`. The last line multiplies what is inside two of the boxes. Note: no quotes around variable names, because quotes would turn them into plain text.

```python
trays = 8          # put a new value in the same box
print(cookies_per_tray * trays)
```

**What happened:** A variable can be overwritten. The old value 5 is gone; the box now holds 8. Variables in a notebook stay alive between cells, so this cell still knows `cookies_per_tray`.

🔗 **Where this appears in Day 1:** `MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"`, `MAX_NEW_TOKENS = 120`, `device = pick_device()` are all variables. Writing names in CAPITALS is just a habit that says “this is a setting you may edit”.

## Lesson 3 — Types: text, whole numbers, decimals, True/False

Every value has a **type**. The four you need: `str` (text, always in quotes), `int` (whole number), `float` (number with a decimal point) and `bool` (only `True` or `False`). `type(x)` tells you the type.

```python
model_name = "Qwen2.5-0.5B"
max_tokens = 120
temperature = 0.7
do_sample = True

print(type(model_name))
print(type(max_tokens))
print(type(temperature))
print(type(do_sample))
```

**What happened:** These four variables are the same kinds of settings used in Day 1. Notice `True` has a capital T and no quotes: it is not text, it is a yes/no value.

```python
print("120" + "5")
print(120 + 5)
print(int("120") + 5)
```

**What happened:** Types change behaviour. Two texts joined with `+` are glued together (`1205`). Two numbers are added (`125`). `int("120")` converts text into a number.

🔗 **Where this appears in Day 1:** `DO_SAMPLE = True` is a bool, `TEMPERATURE = 0.7` is a float, `MAX_NEW_TOKENS = 120` is an int, `MODEL_NAME` is a str. Giving the wrong type (e.g. `"120"` with quotes) causes errors when the model runs.

## Lesson 4 — Maths and f-strings

Python is a calculator: `+ - * /` work as expected, `**` is “to the power of”, `//` is whole-number division. An **f-string** (text with `f` before the quote) lets you put variables inside text using `{ }`.

```python
params = 494_000_000            # underscores just make big numbers readable
bytes_per_param = 2
memory_gb = params * bytes_per_param / 1e9

print(memory_gb)
print(f"The model needs about {memory_gb} GB")
print(f"Rounded: {memory_gb:.1f} GB")
print(f"In millions: {params / 1e6:.0f} million parameters")
```

**What happened:** `1e9` means 1 followed by 9 zeros (one billion). This is the real memory calculation from Day 1: 494 million numbers × 2 bytes each ≈ 1 GB. Inside `{ }`, `:.1f` means “show 1 decimal place” and `:.0f` means “no decimals”.

🔗 **Where this appears in Day 1:** `print(f"Model loaded. Parameters: {n_params / 1e6:.0f} million")` in Day 1 is exactly this pattern. You will use this calculation again on Day 14 for 4-bit models (0.5 bytes per parameter).

## Lesson 5 — Lists: many values in one variable

A **list** holds several values in order, inside square brackets `[ ]`, separated by commas. Each item has a position number called an **index**, and counting starts at **0**, not 1.

```python
prompts = [
    "Explain AI in one line.",
    "Give a tip for cheaper plastic parts.",
    "Write a bakery motto.",
]

print(len(prompts))     # how many items
print(prompts[0])       # first item
print(prompts[2])       # third item
print(prompts[-1])      # -1 means the last item
```

**What happened:** `len(...)` counts the items. `prompts[0]` is the first item because counting starts at 0. Negative numbers count from the end.

```python
prompts.append("What is LoRA?")   # add an item at the end
print(len(prompts))
print(prompts)
```

**What happened:** `.append(...)` adds to the list. The dot means “do this action on this list” (more in Lesson 9).

🔗 **Where this appears in Day 1:** `PROMPTS = [ ... ]` in Day 1 is a list of questions. Training datasets (Day 4) are lists of examples, and the model’s output is a list of token numbers.

## Lesson 6 — Dictionaries: values with labels

A **dictionary** stores pairs of **key → value** inside curly brackets `{ }`. Instead of a position number, you look things up by the key, like a word in a real dictionary.

```python
message = {"role": "user", "content": "Write a bakery motto."}

print(message["role"])
print(message["content"])
```

**What happened:** `"role"` and `"content"` are keys. `message["role"]` looks up the value stored under that key.

```python
settings = {"max_new_tokens": 120, "do_sample": True}
settings["temperature"] = 0.7    # add a new key
settings["max_new_tokens"] = 60  # change an existing key
print(settings)
```

**What happened:** Assigning to a key that doesn’t exist creates it; assigning to one that exists replaces its value.

```python
messages = [
    {"role": "system", "content": "You are helpful."},
    {"role": "user", "content": "Hi!"},
]
print(messages[1]["content"])
```

**What happened:** A list of dictionaries. `messages[1]` picks the second dictionary, then `["content"]` picks its text. Reading left to right: list position first, then key.

🔗 **Where this appears in Day 1:** This is exactly the `messages` chat format in Day 1, and `gen_settings` is a dictionary of generation options. Every training example on Day 4 will be a dictionary like `{"instruction": ..., "response": ...}`.

## Lesson 7 — if / else: making decisions

`if` runs a block of code only when a condition is True. `elif` (else-if) checks another condition; `else` runs when nothing above matched. The lines that belong to the `if` are **indented** (pushed right by 4 spaces). Indentation is how Python knows what belongs together.

```python
gpu_memory_gb = 15

if gpu_memory_gb >= 24:
    print("You can train a 7B model with QLoRA.")
elif gpu_memory_gb >= 12:
    print("Good for 1-3B models.")
else:
    print("Stick to tiny models or CPU.")
```

**What happened:** Python checks top to bottom and runs only the **first** matching block. `>=` means “greater than or equal”. Others: `==` equal (two signs!), `!=` not equal, `<`, `>`. The colon `:` at the end of each condition line is required.

```python
do_sample = False
settings = {"max_new_tokens": 120}
if do_sample:
    settings["temperature"] = 0.7
print(settings)
```

**What happened:** `if do_sample:` means “if this is True”. It was False, so temperature was not added — the same trick Day 1 uses.

🔗 **Where this appears in Day 1:** `pick_device()` in Day 1 is an `if` chain: CUDA GPU? else Intel GPU? else Apple GPU? else CPU. And `if DO_SAMPLE:` adds temperature only when sampling.

## Lesson 8 — for loops: repeating work

A `for` loop runs the indented block once for **each item** in a list. The loop variable (here `prompt`) holds the current item each time round.

```python
prompts = ["Explain AI.", "Bakery motto.", "What is LoRA?"]

for prompt in prompts:
    print("Asking the model:", prompt)
print("All done")
```

**What happened:** The indented line ran 3 times. `print("All done")` is not indented, so it runs once after the loop finishes.

```python
for t in [0.2, 0.7, 1.3]:
    print(f"Experiment with temperature = {t}")

for epoch in range(3):
    print("Training epoch", epoch)
```

**What happened:** `range(3)` produces 0, 1, 2. Training loops (Day 6+) look exactly like this: repeat over epochs, and inside, repeat over batches of data.

🔗 **Where this appears in Day 1:** `for prompt in PROMPTS:` in Day 1 asks the model each question in turn. The experiments loop over temperatures with `for t in [0.2, 0.7, 1.3]:`.

## Lesson 9 — Functions: reusable named blocks

A **function** is a named recipe. You write it once with `def`, then **call** it by name whenever you need it. Values you pass in are **arguments**; `return` hands a result back. Arguments can have **default values** used when you don’t give one.

```python
def memory_gb(params, bytes_per_param=2):
    total = params * bytes_per_param / 1e9
    return total

print(memory_gb(494_000_000))
print(memory_gb(494_000_000, 0.5))
print(memory_gb(7_000_000_000, bytes_per_param=4))
```

**What happened:** `def` defines the recipe; nothing runs until it is called. First call uses the default `bytes_per_param=2`. Second passes 0.5 by position. Third passes it by name (`bytes_per_param=4`), which is clearer and is how most Hugging Face settings are passed.

```python
def pick_device(has_nvidia, has_intel):
    if has_nvidia:
        return "cuda"
    if has_intel:
        return "xpu"
    return "cpu"

print(pick_device(True, False))
print(pick_device(False, True))
print(pick_device(False, False))
```

**What happened:** A simplified copy of Day 1’s `pick_device()`. `return` ends the function immediately, so once a match is found the later lines never run.

🔗 **Where this appears in Day 1:** Day 1 defines `pick_device()`, `main()` and, in the notebook, `ask(prompt, max_new_tokens=120, do_sample=True, temperature=0.7, ...)` — a function with default arguments, exactly like `memory_gb`. Calls such as `from_pretrained(MODEL_NAME, dtype="auto")` pass arguments by name.

## Lesson 10 — Libraries, objects and the dot

A **library** is code someone else wrote that you load with `import`. Libraries give you **objects**: things that hold data *and* have actions. You reach an object’s actions and data with a **dot**: `object.action()` or `object.data`. An action attached to an object is called a **method**.

```python
import math
print(math.sqrt(144))

from statistics import mean
losses = [2.8, 2.1, 1.7, 1.5]
print(mean(losses))
```

**What happened:** `import math` loads the whole library; you then write `math.sqrt`. `from statistics import mean` pulls out just `mean`, so you can call it directly. The second style is how Day 1 imports `AutoTokenizer`.

```python
motto = "  fresh from the oven  "
print(motto.strip())
print(motto.strip().upper())
print(motto.count("o"))
```

**What happened:** A piece of text is an object. `.strip()` removes spaces at the ends; `.upper()` makes capitals. Dots can be **chained**: strip first, then upper. Day 1 ends with `answer.strip()`.

🔗 **Where this appears in Day 1:** `import torch` and `from transformers import AutoTokenizer, AutoModelForCausalLM` load libraries. Then everything uses dots: `AutoTokenizer.from_pretrained(...)`, `model.to(device)`, `model.eval()`, `model.generate(...)`, `tokenizer.decode(...)`, `torch.cuda.is_available()` (a library inside a library).

## Lesson 11 — Slicing: taking part of a list

`list[start:end]` takes a slice from position `start` up to, but **not including**, `end`. Leave one side empty to mean “from the beginning” or “to the end”.

```python
tokens = [101, 2054, 2003, 7393, 102, 3000, 4000, 5000]
prompt_length = 5

print(tokens[0:3])
print(tokens[:prompt_length])     # the prompt part
print(tokens[prompt_length:])     # everything after: the new answer
```

**What happened:** This is the key trick from Day 1. `generate()` returns prompt + answer as one list, so we slice from `prompt_length` to the end to keep only the answer.

```python
batch = [[1, 2, 3], [4, 5, 6]]     # a list of lists = a small table
print(batch[0])
print(batch[0][1:])
print(len(batch), len(batch[0]))
```

**What happened:** Lists can hold lists, giving rows and columns. `batch[0]` is the first row; `batch[0][1:]` slices that row. Model data is shaped like this: [how many sequences, how many tokens]. In PyTorch this size is called the **shape**.

🔗 **Where this appears in Day 1:** `new_tokens = output_ids[0][prompt_length:]` = take sequence 0, then slice off the prompt. `inputs["input_ids"].shape[1]` = the second number of the shape, i.e. how many tokens are in the prompt.

## Lesson 13 — Mini project: build a toy tokenizer

Let’s combine everything into a tiny version of what a real tokenizer does: turn text into numbers and back. Real tokenizers split words into smaller pieces and have ~150,000 entries, but the idea is identical.

```python
text = "the model reads the text"
words = text.split()            # split text at spaces -> list
print(words)

vocab = {}                       # empty dictionary: word -> number
for word in words:
    if word not in vocab:
        vocab[word] = len(vocab) # next free number
print(vocab)
```

**What happened:** `.split()` cuts text into a list of words. The loop gives each new word the next free number. `not in` checks whether a key is missing. Running this creates our **vocabulary**.

```python
def encode(text):
    return [vocab[w] for w in text.split()]

id_to_word = {number: word for word, number in vocab.items()}

def decode(ids):
    return " ".join(id_to_word[i] for i in ids)

ids = encode("the text reads the model")
print(ids)
print(decode(ids))
```

**What happened:** `encode` turns text into IDs; `decode` turns IDs back into text. `[vocab[w] for w in ...]` is a **list comprehension**: a one-line `for` loop that builds a list. `.items()` gives each key-value pair; we flip them to look up words by number. `" ".join(...)` glues words with spaces.

🔗 **Where this appears in Day 1:** This is precisely `tokenizer(text)["input_ids"]` and `tokenizer.decode(ids)` from Day 1. The model only ever sees the numbers. A language model’s job is: given these numbers, predict the next number.
